```python
df = pd.read_excel('Business Sales Data Analysis.xlsx')
```

```python
df.head()
df.shape
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df = df.drop_duplicates()

```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Month'] = df['Order_Date'].dt.month_name()
df['Year'] = df['Order_Date'].dt.year
```

```python
df['Total_Revenue'] = df['Quantity'] * df['Unit_Price']
```

```python
df['Discount_Amount'] = (
    df['Total_Revenue'] * df['Discount_Percent'] / 100
)
```

```python
display(df.head())
```

```python
total_sales = df['Sales_Amount'].sum()

print("Total Sales: ", round(total_sales, 2))
```

```python
total_transactions = df['Transaction_ID'].nunique()

print("Total Transactions:", total_transactions)
```

```python
total_quantity = df['Quantity'].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
aov = total_sales / total_transactions

print("Average Order Value: ₹", round(aov, 2))
```

```python
total_customers = df['Customer_ID'].nunique()

print("Total Customers:", total_customers)
```

```python
total_discount = (
    df['Quantity'] * df['Unit_Price'] *
    df['Discount_Percent'] / 100
).sum()

print("Total Discount: ₹", round(total_discount, 2))
```

```python
kpi = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Transactions',
        'Total Quantity Sold',
        'Average Order Value',
        'Total Customers',
        'Total Discount'
    ],
    'Value': [
        round(total_sales, 2),
        total_transactions,
        total_quantity,
        round(aov, 2),
        total_customers,
        round(total_discount, 2)
    ]
})

display(kpi)
```

```python

category_sales = df.groupby('Category')['Sales_Amount'].sum()

category_sales = category_sales.sort_values(ascending=False)

plt.figure(figsize=(10,5))

plt.bar(category_sales.index, category_sales.values)

plt.title('Category-wise Sales Analysis')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby('Product_Name')['Sales_Amount'].sum()

top_products = product_sales.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

plt.bar(top_products.index, top_products.values)

plt.title('Top 10 Products by Sales')
plt.xlabel('Product Name')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

city_sales = df.groupby('City')['Sales_Amount'].sum()

top_cities = city_sales.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

plt.bar(top_cities.index, top_cities.values)

plt.title('Top 10 Cities by Sales')
plt.xlabel('City')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

df['Order_Date'] = pd.to_datetime(df['Order_Date'])

df['Month'] = df['Order_Date'].dt.month

monthly_sales = df.groupby('Month')['Sales_Amount'].sum()

plt.figure(figsize=(12,6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(range(1,13))
plt.grid()
plt.tight_layout()
plt.show()
```

```python

payment_sales = df.groupby('Payment_Method')['Sales_Amount'].sum()

plt.figure(figsize=(8,8))

plt.pie(
    payment_sales.values,
    labels=payment_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution by Payment Method')
plt.show()
```

```python
import plotly.express as px
import pandas as pd

print("Plotly imported successfully")
```

```python

category_sales = df.groupby('Category', as_index=False)['Sales_Amount'].sum()

fig = px.bar(
    category_sales,
    x='Category',
    y='Sales_Amount',
    title='Category-wise Sales Analysis',
    color='Category',
    text_auto='.2s'
)

fig.show()
```

```python
product_sales = df.groupby(
    'Product_Name', as_index=False
)['Sales_Amount'].sum()

top_products = product_sales.sort_values(
    by='Sales_Amount',
    ascending=False
).head(10)

fig = px.bar(
    top_products,
    x='Product_Name',
    y='Sales_Amount',
    title='Top 10 Products by Sales',
    color='Sales_Amount',
    text_auto='.2s'
)

fig.update_layout(xaxis_tickangle=-45)

fig.show()
```

```python
city_sales = df.groupby(
    'City', as_index=False
)['Sales_Amount'].sum()

top_cities = city_sales.sort_values(
    by='Sales_Amount',
    ascending=False
).head(10)

fig = px.bar(
    top_cities,
    x='City',
    y='Sales_Amount',
    title='Top 10 Cities by Sales',
    color='Sales_Amount',
    text_auto='.2s'
)

fig.show()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

df['Month'] = df['Order_Date'].dt.to_period('M').astype(str)

monthly_sales = df.groupby(
    'Month', as_index=False
)['Sales_Amount'].sum()

fig = px.line(
    monthly_sales,
    x='Month',
    y='Sales_Amount',
    title='Monthly Sales Trend',
    markers=True
)

fig.show()
```

```python
payment_sales = df.groupby(
    'Payment_Method', as_index=False
)['Sales_Amount'].sum()

fig = px.pie(
    payment_sales,
    names='Payment_Method',
    values='Sales_Amount',
    title='Sales Distribution by Payment Method',
    hole=0.4
)

fig.show()
```

```python

# Business Insights Summary

insights = {
    'Analysis': [
        'Total Sales',
        'Top Category',
        'Top Product',
        'Top City',
        'Best Payment Method',
        'Best Sales Channel',
        'Best Sales Month'
    ],

    'Result': [
        df['Sales_Amount'].sum(),
        df.groupby('Category')['Sales_Amount'].sum().idxmax(),
        df.groupby('Product_Name')['Sales_Amount'].sum().idxmax(),
        df.groupby('City')['Sales_Amount'].sum().idxmax(),
        df.groupby('Payment_Method')['Sales_Amount'].sum().idxmax(),
        df.groupby('Sales_Channel')['Sales_Amount'].sum().idxmax(),
        df.groupby(df['Order_Date'].dt.month)['Sales_Amount'].sum().idxmax()
    ]
}

insights_df = pd.DataFrame(insights)

display(insights_df)
```
